# Absenteeism — Phase 3: Integration / Deployment

**Purpose:** load the trained model through the `absenteeism_module`, score unseen data,
and export predictions for Tableau.

**Inputs:** `absenteeism_module.py`, `model`, `scaler`, `Absenteeism_new_data.csv`
**Output:** `Absenteeism_predictions.csv`

This notebook contains almost no logic — that is the point. All the preprocessing and
scoring is encapsulated in the module, so deployment is just three calls.

## 1. Import the module

```
from absenteeism_module import AbsenteeismModel
```

## 2. Load the model and score new data

```
model = AbsenteeismModel('model', 'scaler')
model.load_and_clean_data('Absenteeism_new_data.csv')
```

## 3. Get the predictions

```
# full table with Probability (of excessive absenteeism) and Prediction (0/1)
model.predicted_outputs().head()
```

```
# you can also get the pieces separately if needed
model.predicted_probability()[:10]
```

```
model.predicted_output_category()[:10]
```

## 4. Export for Tableau

```
model.predicted_outputs().to_csv('Absenteeism_predictions.csv', index=False)
print('Predictions exported.')
```

## Summary

The module turns scoring new data into a three-line operation. The exported CSV carries
each employee's probability and predicted class alongside their features, ready for the
Tableau dashboards in Phase 4.

**Note:** the scaler is only ever *transformed* on new data, never re-fitted — the
fitted parameters come from the training set, pickled in Phase 2.
